```python
with open('occ_codebook.txt', 'r', encoding='utf-8') as oc:
    lines = oc.read().splitlines()
```

```python
occ_dict = {}

current_code = None

for line in lines:
    line = line.strip()

    if line == "":
        continue

    # if it's numeric, treat as code candidate
    if line.isdigit():
        current_code = line.zfill(4)   # normalize EVERYTHING to 4-digit
    else:
        # treat as label
        if current_code is not None:
            occ_dict[current_code] = line
            current_code = None
```

```python

df = pd.read_csv('cps_00001.csv.gz')
df = df[['YEAR', 'REGION', 'STATEFIP', 'COUNTY', 'METRO', 'ASECWT', 'AGE', 'SEX',
       'RACE', 'OCC', 'UHRSWORK1', 'EDUC', 'SCHLCOLL', 'INCWAGE', 'OFFTOTVAL',
       'POVERTY']]

occ_dict = {}

current_code = None

for line in lines:
    line = line.strip()

    if line == "":
        continue

    # if it's numeric, treat as code candidate
    if line.isdigit():
        current_code = line.zfill(4)   # normalize EVERYTHING to 4-digit
    else:
        # treat as label
        if current_code is not None:
            occ_dict[current_code] = line

df["OCC"] = df["OCC"].astype(str).str.zfill(4)
df["OCC_LABEL"] = df["OCC"].map(occ_dict)          


ubi_subset = df[['EDUC','SCHLCOLL','INCWAGE','AGE','SEX','RACE','OCC','OCC_LABEL']]

race_map = {
    100: 'White',
    200: 'Black',
    300: 'American Indian',
    651: 'Asian Indian',
    652: 'Chinese',
    801: 'White & Black',
    802: 'White & AIAN',
    803: 'White & Asian',
    804: 'Black & AIAN',
    805: 'Black & Asian',
    806: 'AIAN & Asian'
}
ubi_subset['RACE_LABEL'] = ubi_subset['RACE'].map(race_map)

sex_map = {
    1: 'Male',
    2: 'Female'
}

ubi_subset['SEX_LABEL'] = ubi_subset['SEX'].map(sex_map)

ubi_subset['INCWAGE'] = ubi_subset['INCWAGE'].replace(99999999, np.nan)
ubi_subset = ubi_subset.dropna(subset=['INCWAGE'])
ubi_subset = ubi_subset[ubi_subset['INCWAGE'] > 0]

result = ubi_subset.groupby(
    ['RACE_LABEL', 'SEX_LABEL']
)['INCWAGE'].median().reset_index()

result = result.sort_values('INCWAGE', ascending=False)
result


```
